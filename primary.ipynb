```python
# Group by 'Province' and calculate the mean of selected numerical columns
grouped = df.groupby("Province")[["PupilTeach", "ClassrmRat", "ToiletRati"]].mean()

# Sort by Pupil-to-Teacher ratio for clarity
grouped = grouped.sort_values("PupilTeach", ascending=False)

# Display the grouped summary
grouped
```

```python
# Check for provinces with extremely high or low values
grouped.style.highlight_max(color='lightcoral').highlight_min(color='lightgreen')
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["ToiletRati"], bins=30, kde=True, color="skyblue")
plt.title("Distribution of Pupil-to-Toilet Ratio")
plt.xlabel("Pupils per Toilet")
plt.ylabel("Number of Schools")
plt.tight_layout()
plt.show()

```

```python

province_avg = df.groupby("Province")["PupilTeach"].mean().sort_values()

plt.figure(figsize=(12, 6))
sns.barplot(x=province_avg.values, y=province_avg.index, palette="viridis")
plt.title("Average Pupil-to-Teacher Ratio by Province")
plt.xlabel("Pupil-to-Teacher Ratio")
plt.ylabel("Province")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="PupilTeach", y="ClassrmRat", hue="Province", alpha=0.6)
plt.title("Pupil-to-Teacher vs Classroom Ratio")
plt.xlabel("Pupil-to-Teacher Ratio")
plt.ylabel("Pupil-to-Classroom Ratio")
plt.legend(title="Province", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
